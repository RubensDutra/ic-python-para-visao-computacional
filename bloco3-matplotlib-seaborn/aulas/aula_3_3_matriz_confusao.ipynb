## 📖 Tabela de Termos

| Termo | Conceito | Função no código |
|-------|---------|------------------|
| **matriz de confusão** | Tabela que mostra acertos e erros por classe | `np.array([[VP_neg, FP_pos],[FN_neg, VP_pos]])` |
| **VP (Verdadeiro Positivo)** | Real sem capacete → modelo disse sem capacete ✅ | `matriz[1][1]` |
| **FP (Falso Positivo)** | Real com capacete → modelo disse sem capacete ❌ | `matriz[0][1]` |
| **FN (Falso Negativo)** | Real sem capacete → modelo disse com capacete ❌ | `matriz[1][0]` |
| **VN (Verdadeiro Negativo)** | Real com capacete → modelo disse com capacete ✅ | `matriz[0][0]` |
| **Precision** | Dos que o modelo sinalizou como infrator, quantos realmente eram? | `VP / (VP + FP)` |
| **Recall** | Dos infratores reais, quantos o modelo encontrou? | `VP / (VP + FN)` |
| **Seaborn** | Biblioteca de visualização baseada no Matplotlib | `import seaborn as sns` |
| **heatmap()** | Gráfico de calor — escurece células com valores maiores | `sns.heatmap(matriz, annot=True)` |
| **annot=True** | Exibe o número dentro de cada célula do heatmap | parâmetro de `sns.heatmap()` |

## 🧠 Conceito

Depois de treinar o YOLO, você precisa saber **se ele está detectando bem**.  
A **matriz de confusão** mostra, para cada classe real, o que o modelo previu.

```
                  PREVISTO PELO MODELO
                  com capacete | sem capacete
REAL  com capacete |    VN      |     FP      |
      sem capacete |    FN      |     VP      |
```

- **Diagonal principal** (VN e VP) → **acertos** — quanto mais escuro, melhor
- **Fora da diagonal** (FP e FN) → **erros**

### Por que o Recall é mais importante neste projeto?

> Um **FN** (Falso Negativo) significa que um motociclista **sem capacete** passou despercebido.  
> O modelo viu, mas classificou como "com capacete" — um infrator real foi ignorado.  
> Esse é o erro mais grave: a fiscalização deixa de agir onde deveria.

Por isso priorizamos o **Recall** — ele mede exatamente quantos infratores reais foram capturados:

```
Recall = VP / (VP + FN)
```

> No projeto real, o YOLO gera a matriz de confusão automaticamente após a avaliação (`val.py`).

````python
# Importar bibliotecas
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
````

## 💻 Código

````python
# ── 1. Definir a matriz de confusão ──────────────────────────────────────────
# Linha = classe REAL | Coluna = classe PREVISTA pelo modelo
# Ordem das classes: [com capacete, sem capacete]
#
#                  Previsto: com   Previsto: sem
# Real: com capacete  [ 90,          5  ]   → 90 acertos, 5 erros (FP)
# Real: sem capacete  [  3,         12  ]   → 3 erros (FN), 12 acertos

matriz = np.array([
    [90,  5],   # linha 0: real = com capacete
    [ 3, 12],   # linha 1: real = sem capacete
])
````

````python
# ── 2. Visualizar com heatmap do Seaborn ─────────────────────────────────────
fig, ax = plt.subplots(figsize=(6, 5))

sns.heatmap(
    matriz,
    annot=True,                                      # exibe números nas células
    fmt='d',                                         # formato inteiro (sem decimais)
    cmap='Blues',                                    # escala de azul: mais escuro = maior valor
    xticklabels=['Com capacete', 'Sem capacete'],    # rótulos das colunas (previsto)
    yticklabels=['Com capacete', 'Sem capacete'],    # rótulos das linhas (real)
    ax=ax
)

ax.set_xlabel('Previsto pelo modelo', fontsize=12)
ax.set_ylabel('Real', fontsize=12)
ax.set_title('Matriz de Confusão — Detecção de capacetes', fontsize=13, weight='bold')

plt.tight_layout()
plt.show()
````

````python
# ── 3. Extrair os valores da matriz ──────────────────────────────────────────
VN = matriz[0][0]   # real com capacete → previsto com capacete (acerto)
FP = matriz[0][1]   # real com capacete → previsto sem capacete (erro)
FN = matriz[1][0]   # real sem capacete → previsto com capacete (erro GRAVE)
VP = matriz[1][1]   # real sem capacete → previsto sem capacete (acerto)

print(f"VN (acertos com capacete): {VN}")
print(f"FP (falso alarme):         {FP}")
print(f"FN (infrator ignorado):    {FN}  ← erro mais grave")
print(f"VP (infrator capturado):   {VP}")
````

````python
# ── 4. Calcular Precision e Recall ───────────────────────────────────────────
precision = VP / (VP + FP)   # dos sinalizados, quantos são reais infratores?
recall    = VP / (VP + FN)   # dos infratores reais, quantos foram pegos?

total   = matriz.sum()
acertos = VN + VP

print(f"\nTotal de detecções: {total}")
print(f"Acertos:            {acertos} ({acertos/total*100:.1f}%)")
print(f"Erros:              {total - acertos} ({(total-acertos)/total*100:.1f}%)")
print(f"\nPrecision: {precision:.2f}  → de cada 10 alarmes, {precision*10:.0f} são reais")
print(f"Recall:    {recall:.2f}  → o modelo captura {recall*100:.0f}% dos infratores reais")
````
