# Projeto Integrador — Bloco 2: Pandas

**Projeto:** IC-FAPEMA — Detecção de capacetes em motociclistas via drone  
**Objetivo:** Pipeline completo de análise de detecções usando Pandas

---

### Etapas do pipeline
1. Criar o DataFrame com detecções do voo
2. Inspecionar a estrutura dos dados
3. Filtrar detecções por classe e posição
4. Agrupar e contar por frame
5. Calcular estatísticas por classe
6. Gerar resumo final do voo

```python
import pandas as pd
```

## 1. Criar o DataFrame

```python
# DataFrame com detecções do voo do drone
# classe 0 = com capacete | classe 1 = sem capacete
voo = pd.DataFrame({
    'frame':    [1, 1, 2, 2, 3, 3],
    'classe':   [1, 0, 1, 0, 1, 1],
    'x_centro': [0.52, 0.75, 0.30, 0.80, 0.45, 0.90],
    'y_centro': [0.38, 0.60, 0.45, 0.25, 0.70, 0.55],
    'largura':  [0.20, 0.18, 0.22, 0.15, 0.19, 0.25],
    'altura':   [0.15, 0.20, 0.17, 0.12, 0.21, 0.18],
})
voo
```

## 2. Inspecionar a estrutura

```python
# Shape: quantas detecções e quantas colunas
shape = voo.shape
print("Shape do Frame:", shape)

# Tipos de dado de cada coluna
dtypes = voo.dtypes
print("\nDtypes:\n", dtypes)

# Primeiras 5 linhas
head = voo.head()
print("\nHead:\n", head)
```

## 3. Filtrar detecções

```python
# Filtro 1: apenas com capacete (classe 0)
sem_capacete = voo[voo['classe'] == 0]
print("Com capacete:\n", sem_capacete)

# Filtro 2: detecções na parte inferior do frame (mais perto do drone)
parte_inferior = voo[voo['y_centro'] > 0.5]
print("\nParte inferior do frame:\n", parte_inferior)
```

## 4. Agrupar por frame

```python
# Quantas detecções aconteceram em cada frame?
frame = voo.groupby('frame').count()
print("Detecções por frame:\n", frame)
```

## 5. Estatísticas por classe

```python
# Média de posição (x_centro e y_centro) por classe
media = voo.groupby('classe').mean()
print("Média por classe:\n", media)

# Média e máximo da altura da bounding box por classe
altura = voo.groupby('classe').agg({'altura': ['mean', 'max']})
print("\nEstatísticas de altura por classe:\n", altura)
```

## 6. Resumo final do voo

```python
# Distribuição por classe
resumo = voo['classe'].value_counts()
print("Distribuição por classe:\n", resumo)

# Contagem explícita
com = (voo['classe'] == 0).sum()
sem = (voo['classe'] == 1).sum()
total = voo.shape[0]

print(f"\n--- Resumo do voo ---")
print(f"Total de detecções : {total}")
print(f"Com capacete (0)   : {com}")
print(f"Sem capacete (1)   : {sem}")
print(f"Taxa de infração   : {sem/total*100:.1f}%")
```

---
## ✅ Pipeline completo — Bloco 2

| Etapa | Função usada | O que fez |
|---|---|---|
| Criar dados | `pd.DataFrame()` | Organizou as detecções em tabela |
| Inspecionar | `shape`, `dtypes`, `head()` | Verificou estrutura e tipos |
| Filtrar | `df[condição]`, `loc` | Isolou classes e regiões do frame |
| Agrupar | `groupby().count()` | Contou detecções por frame |
| Estatísticas | `groupby().mean()`, `agg()` | Calculou médias e máximos por classe |
| Resumo | `value_counts()`, `.sum()` | Gerou relatório final do voo |
