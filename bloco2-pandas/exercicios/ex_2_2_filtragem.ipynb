**Projeto:** IC-FAPEMA — Detecção de capacetes em motociclistas via drone  
**Objetivo:** Praticar filtragem de detecções do drone por classe e posição

---

```python
import pandas as pd
```

```python
# DataFrame do voo 01 — mesmo da aula anterior
# classe 0 = com capacete | classe 1 = sem capacete
imagens = pd.DataFrame({
    'frame':    [0, 1, 2, 3, 4, 5],
    'classe':   [1, 1, 0, 0, 1, 1],
    'x_centro': [0.52, 0.36, 0.78, 0.95, 0.36, 0.75],
    'y_centro': [0.33, 0.44, 0.01, 0.74, 0.25, 0.35],
    'largura':  [0.52, 0.15, 0.10, 0.22, 0.66, 0.75],
    'altura':   [0.30, 0.46, 0.88, 0.24, 0.40, 0.25]
})
```

```python
# 1. Primeiras 3 detecções
imagens.head(3)
```

```python
# 2. Apenas sem capacete (classe 1)
sem_capacetes = imagens.loc[imagens['classe'] == 1]
sem_capacetes
```

```python
# 3. Detecções na parte inferior do frame (y_centro > 0.4)
# y maior = mais abaixo na imagem
deteccao = imagens.loc[imagens['y_centro'] > 0.4]
deteccao
```

```python
# 4. Contagem por classe
contagem = imagens['classe'].value_counts()
contagem
```

## Resumo do exercício

| O que praticou | Função no código | Para que serve no projeto |
|---|---|---|
| Ver primeiras linhas | `head(3)` | Inspecionar início das detecções |
| Filtrar sem capacete | `loc[classe == 1]` | Isolar infratores |
| Filtrar por posição | `loc[y_centro > 0.4]` | Selecionar região do frame |
| Contar por classe | `value_counts()` | Ver distribuição das detecções |
