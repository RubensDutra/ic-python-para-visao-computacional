```python
import pandas as pd
```

## Criando o DataFrame com detecções do voo 01

```python
# DataFrame com 6 detecções de um voo do drone
# classe 0 = com capacete | classe 1 = sem capacete
voo_01 = pd.DataFrame({
    'frame':    [0, 1, 2, 3, 4, 5],
    'classe':   [1, 0, 0, 0, 1, 1],
    'x_centro': [0.52, 0.36, 0.78, 0.95, 0.36, 0.75],
    'y_centro': [0.33, 0.44, 0.01, 0.74, 0.25, 0.35],
    'largura':  [0.52, 0.15, 0.10, 0.22, 0.66, 0.75],
    'altura':   [0.30, 0.46, 0.88, 0.24, 0.40, 0.25]
})

# 1. Tabela completa
print("Tabela:\n", voo_01)
```

```python
# 2. Shape: (linhas, colunas)
print("Formato da estrutura", voo_01.shape)
```

```python
# 3. Coluna frame como Series
print("Quadro de detecção:", voo_01['frame'])
```

## Resumo do exercício

| O que praticou | Função no código | Para que serve no projeto |

| Ver estrutura | `voo_01.shape` | Saber quantas detecções e colunas existem |
| Extrair Series | `voo_01['frame']` | Acessar uma coluna específica |
