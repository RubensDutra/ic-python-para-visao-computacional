# Exercício 2.3 — Agrupamentos e Estatísticas

**Bloco 2 — Pandas**  
**Projeto:** IC-FAPEMA — Detecção de capacetes em motociclistas via drone  
**Objetivo:** Praticar groupby e estatísticas descritivas nas detecções do drone

---

```python
import pandas as pd
```

```python
# DataFrame do exercício
# classe 0 = com capacete | classe 1 = sem capacete
deteccoes = pd.DataFrame({
    'frame':    [1, 1, 2, 2, 3, 3],
    'classe':   [1, 1, 0, 0, 0, 0],
    'x_centro': [0.52, 0.75, 0.30, 0.80, 0.45, 0.90],
    'y_centro': [0.38, 0.60, 0.45, 0.25, 0.70, 0.55],
    'largura':  [0.20, 0.18, 0.22, 0.15, 0.19, 0.25],
    'altura':   [0.15, 0.20, 0.17, 0.12, 0.21, 0.18],
})
```

```python
# 1. Média de x_centro e y_centro agrupada por classe
estatistica = deteccoes.groupby('classe').agg({'x_centro': ['mean']})
estatistica
```

```python
# 2. Contagem de detecções por frame
contagem = deteccoes.groupby('frame').count()
contagem
```

```python
# 3. Média e máximo da coluna altura agrupados por classe
coluna_altura = deteccoes.groupby('classe').agg({'altura': ['mean', 'max']})
coluna_altura
```

## Resumo do exercício

| O que praticou | Função no código | Para que serve no projeto |
|---|---|---|
| Média por classe | `groupby('classe').agg({'x_centro': ['mean']})` | Posição média por tipo de detecção |
| Contagem por frame | `groupby('frame').count()` | Quantas detecções por frame do voo |
| Média e máximo | `groupby('classe').agg({'altura': ['mean', 'max']})` | Tamanho das bounding boxes por classe |
