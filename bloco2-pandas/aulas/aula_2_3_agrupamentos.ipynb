## 📖 Tabela de Termos

| Termo | Conceito | Função no código |
|-------|---------|------------------|
| **groupby()** | Agrupa linhas por valor de uma coluna | `df.groupby('classe')` |
| **mean()** | Calcula a média dos valores do grupo | `df.groupby('classe')['altura'].mean()` |
| **sum()** | Soma os valores do grupo | `df.groupby('classe').sum()` |
| **count()** | Conta o número de linhas por grupo | `df.groupby('frame').count()` |
| **agg()** | Aplica várias estatísticas de uma vez | `df.groupby('classe').agg({'altura': ['mean', 'max']})` |

## 🧠 Conceito

Após filtrar as detecções, precisamos **resumir os dados** para tirar conclusões do voo:

- *Qual classe tem bounding boxes maiores em média?*
- *Quantas detecções aconteceram em cada frame?*
- *Qual o tamanho máximo de uma detecção por classe?*

O `groupby()` funciona em 3 etapas:
1. **Separa** as linhas por grupo (ex: classe 0 e classe 1)
2. **Aplica** a função estatística em cada grupo
3. **Combina** os resultados em uma tabela

```python
import pandas as pd
```

## 💻 Código

```python
# DataFrame com detecções do drone
# classe 0 = com capacete | classe 1 = sem capacete
deteccoes = pd.DataFrame({
    'frame':    [1, 1, 2, 2, 3, 3],
    'classe':   [1, 0, 1, 0, 0, 1],
    'x_centro': [0.52, 0.75, 0.30, 0.80, 0.45, 0.90],
    'y_centro': [0.38, 0.60, 0.45, 0.25, 0.70, 0.55],
    'largura':  [0.20, 0.18, 0.22, 0.15, 0.19, 0.25],
    'altura':   [0.15, 0.20, 0.17, 0.12, 0.21, 0.18],
})
deteccoes
```

```python
# Média de x_centro e y_centro agrupada por classe
# Pergunta: qual classe aparece mais no centro ou nas bordas do frame?
media_por_classe = deteccoes.groupby('classe')[['x_centro', 'y_centro']].mean()
print("Média de posição por classe:")
media_por_classe
```

```python
# Contagem de detecções por frame
# Pergunta: em qual frame o drone detectou mais motociclistas?
contagem_por_frame = deteccoes.groupby('frame').count()
print("Detecções por frame:")
contagem_por_frame
```

```python
# Média e máximo da coluna altura agrupados por classe
# agg() permite calcular várias estatísticas de uma vez
estatisticas_altura = deteccoes.groupby('classe').agg({'altura': ['mean', 'max']})
print("Estatísticas de altura por classe:")
estatisticas_altura
```
