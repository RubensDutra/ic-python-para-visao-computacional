## 📖 Tabela de Termos

| Termo | Conceito | Função no código |
|-------|---------|------------------|
| **DataFrame** | Tabela com linhas e colunas | `pd.DataFrame({...})` |
| **Series** | Uma única coluna do DataFrame | `deteccoes['classe']` |
| **índice** | Número da linha (começa em 0) | `deteccoes.index` |
| **coluna** | Cada campo da tabela | `deteccoes.columns` |
| **dtype** | Tipo de dado de cada coluna | `deteccoes.dtypes` |
| **shape** | Quantidade de linhas e colunas | `deteccoes.shape` |

## 🧠 Conceito

No projeto, cada detecção do drone gera uma linha de dados:

> *"Frame 42 — motociclista na posição X, Y — sem capacete"*

O **DataFrame** organiza todas essas detecções em uma tabela:

- Cada **linha** = uma detecção
- Cada **coluna** = um atributo da detecção
- Uma **Series** = só a coluna `classe`, por exemplo

```python
import pandas as pd
```

## 💻 Código — Criando o DataFrame de detecções

```python
# Criando um DataFrame com detecções do drone
# Cada linha = um motociclista detectado
# classe 0 = com capacete | classe 1 = sem capacete
deteccoes = pd.DataFrame({
    'frame':    [1, 1, 2, 2, 3],
    'classe':   [1, 0, 1, 0, 0],
    'x_centro': [0.52, 0.75, 0.30, 0.80, 0.45],
    'y_centro': [0.38, 0.60, 0.45, 0.25, 0.70],
    'largura':  [0.20, 0.18, 0.22, 0.15, 0.19],
    'altura':   [0.15, 0.20, 0.17, 0.12, 0.21]
})

# Visualizar a tabela completa
print(deteccoes)
```

```python
# Shape: (linhas, colunas)
# Retorna (5, 6) — 5 detecções, 6 atributos
print("Shape:", deteccoes.shape)
```

```python
# Tipos de dado de cada coluna
print("Tipos:\n", deteccoes.dtypes)
```

```python
# Acessar uma coluna — isso retorna uma Series
print("Coluna classe:\n", deteccoes['classe'])
```

```python
# Ver os nomes das colunas
print("Colunas:", deteccoes.columns.tolist())

# Ver o índice (número das linhas)
print("Índice:", deteccoes.index.tolist())
```
