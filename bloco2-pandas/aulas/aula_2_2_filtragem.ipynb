## 📖 Tabela de Termos

| Termo | Conceito | Função no código |
|-------|---------|------------------|
| **filtro booleano** | Seleciona linhas com base em uma condição | `df[df['classe'] == 1]` |
| **loc** | Acessa linhas/colunas por rótulo ou condição | `df.loc[df['x_centro'] > 0.5]` |
| **iloc** | Acessa linhas/colunas por posição numérica | `df.iloc[0]` |
| **head()** | Mostra as primeiras N linhas | `df.head(3)` |
| **value_counts()** | Conta quantas vezes cada valor aparece | `df['classe'].value_counts()` |

## 🧠 Conceito

Após um voo, o drone gera centenas de detecções. Precisamos **filtrar** para responder perguntas como:

- *Quantos motociclistas estão sem capacete?*
- *Quais detecções estão na metade direita do frame?*
- *Qual a distribuição entre com e sem capacete?*

```python
import pandas as pd
```

## 💻 Código

```python
# DataFrame com detecções do drone
# classe 0 = com capacete | classe 1 = sem capacete
deteccoes = pd.DataFrame({
    'frame':    [1, 1, 2, 2, 3, 3],
    'classe':   [1, 0, 1, 0, 0, 1],
    'x_centro': [0.52, 0.75, 0.30, 0.80, 0.45, 0.90],
    'y_centro': [0.38, 0.60, 0.45, 0.25, 0.70, 0.55],
})
```

```python
# Primeiras 3 linhas
deteccoes.head(3)
```

```python
# Filtro: apenas sem capacete (classe 1)
# Retorna só as linhas onde classe == 1
sem_capacete = deteccoes[deteccoes['classe'] == 1]
print("Sem capacete:\n", sem_capacete)
```

```python
# Filtro: detecções na metade direita do frame (x > 0.5)
# x_centro > 0.5 significa que o objeto está na metade direita
direita = deteccoes[deteccoes['x_centro'] > 0.5]
print("Metade direita:\n", direita)
```

```python
# Contagem por classe
# Mostra quantas detecções existem de cada classe
print("Contagem por classe:\n", deteccoes['classe'].value_counts())
```

```python
# Acessar a primeira detecção pelo índice numérico
print("Primeira detecção:\n", deteccoes.iloc[0])
```
