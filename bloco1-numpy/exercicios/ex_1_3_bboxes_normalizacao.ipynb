# Exercício 1.3 — Bounding Boxes e Normalização

**Bloco 1 — NumPy**  
**Projeto:** IC-FAPEMA — Detecção de capacetes em motociclistas via drone  
**Objetivo:** Praticar normalização de imagem e manipulação de anotações YOLO

---

```python
import numpy as np
```

## Parte A — Normalização da imagem

```python
# Frame bruto do drone — 480x640 colorido, uint8 (0 a 255)
imagem = np.random.randint(0, 256, size=(480, 640, 3), dtype=np.uint8)

# Passo 1: converte para float32 (necessário antes de dividir)
imagem_float = imagem.astype(np.float32)

# Passo 2: normaliza — todos os pixels ficam entre 0.0 e 1.0
imagem_normalizada = imagem_float / 255.0

print("dtype original:", imagem.dtype)              # uint8
print("dtype normalizado:", imagem_normalizada.dtype) # float32
print("Valor máximo:", imagem_normalizada.max())      # 1.0
print("Valor mínimo:", imagem_normalizada.min())      # 0.0
```

## Parte B — Anotações no formato YOLO

```python
# Array de anotações — cada linha é um motociclista detectado
# Formato: [classe, x_centro, y_centro, largura, altura]
# classe 1 = sem capacete | classe 0 = com capacete
anotacoes = np.array([
    [1, 0.75, 0.85, 0.97, 0.15],   # sem capacete
    [1, 0.05, 0.85, 0.69, 0.25],   # sem capacete
    [1, 0.55, 0.85, 0.37, 0.35],   # sem capacete
    [0, 0.35, 0.85, 0.34, 0.45],   # com capacete
], dtype=np.float32)

# Shape do array: (4 objetos, 5 valores por objeto)
print("Shape:", anotacoes.shape)       # (4, 5)

# Coluna 0 → classes de todos os objetos
print("Classes:", anotacoes[:, 0])     # [1. 1. 1. 0.]

# Colunas 1 a 4 → bboxes completas (x_centro, y_centro, largura, altura)
print("Centros:\n", anotacoes[:, 1:5])
```

---
## Resumo do exercício

| O que praticou | Para que serve no projeto |
|---|---|
| `astype(np.float32)` | Preparar imagem antes de normalizar |
| `/ 255.0` | Normalizar pixels para 0.0–1.0 (entrada no modelo) |
| Array de anotações YOLO | Representar 4 detecções com classe e posição |
| `anotacoes[:, 0]` | Extrair classes de todos os objetos |
| `anotacoes[:, 1:5]` | Extrair bboxes completas de todos os objetos |
