**Questões:**
1. Crie uma imagem colorida de resolução **720×1280** com pixels aleatórios em `uint8`
2. Imprima o `shape`, o `dtype` e o total de valores (`size`)
3. Acesse o pixel da linha 100, coluna 200 e imprima seus valores RGB
4. Qual a diferença entre `shape` e `size` nesse caso?

```python
# Importa o NumPy
import numpy as np
```

## Resolução do aluno (Rubens Dutra)

```python
# Questões 1, 2 e 3 — Imagem 1 (720x1280)

# Cria a imagem com pixels aleatórios
# size=(720, 1280, 3) → altura=720, largura=1280, 3 canais RGB
# dtype=uint8 → valores de pixel entre 0 e 255
frame_imagem1 = np.random.randint(0, 256, size=(720, 1280, 3), dtype=np.uint8)

# Inspeciona a imagem
print("Shape: ", frame_imagem1.shape)         # (720, 1280, 3)
print("dtype: ", frame_imagem1.dtype)         # uint8
print("Total de valores (size): ", frame_imagem1.size)  # 2.764.800

# Acessa o pixel na linha 100, coluna 200
# Forma preferida no NumPy: [linha, coluna] com vírgula
print("Pixel em [100, 200]:", frame_imagem1[100, 200])  # ex: [173  42 201]

# Acessando cada canal separadamente
print("Canal R (vermelho):", frame_imagem1[100, 200, 0])
print("Canal G (verde):  ", frame_imagem1[100, 200, 1])
print("Canal B (azul):   ", frame_imagem1[100, 200, 2])
```

```python
# Questão 4 — Diferença entre shape e size

# shape → estrutura da imagem: (altura, largura, canais)
# size  → total de valores numéricos: altura x largura x canais
#
# Exemplo para 720x1280:
# shape = (720, 1280, 3)  → mostra a estrutura
# size  = 2.764.800       → 720 x 1280 x 3 valores individuais
#
# Pixels = 720 x 1280 = 921.600 pontos na imagem
# Cada pixel tem 3 valores (R, G, B) → por isso size é 3x maior que pixels

pixels = frame_imagem1.shape[0] * frame_imagem1.shape[1]
print("Total de pixels:", pixels)              # 921.600
print("Total de valores (size):", frame_imagem1.size)  # 2.764.800
```

```python
# Questão 5 — Segunda imagem com resolução diferente

# Cria imagem com resolução menor para comparar
# size=(200, 201, 3) → altura=200, largura=201, 3 canais RGB
frame_imagem2 = np.random.randint(0, 256, size=(200, 201, 3), dtype=np.uint8)

print("Shape (forma): ", frame_imagem2.shape)   # (200, 201, 3)
print("dtype: ", frame_imagem2.dtype)           # uint8
print("Total de valores: ", frame_imagem2.size) # 120.600

# Acessa pixel na linha 100, coluna 200
print("Pixel [100, 200]:", frame_imagem2[100, 200])  # [R G B]
```
