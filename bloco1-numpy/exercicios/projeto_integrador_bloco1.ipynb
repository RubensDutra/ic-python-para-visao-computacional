# Projeto Integrador — Bloco 1

**Bloco 1 — NumPy**  
**Projeto:** IC-FAPEMA — Detecção de capacetes em motociclistas via drone  
**Objetivo:** Pipeline completo de pré-processamento — do frame bruto até as anotações prontas para o modelo

---

## O que este código faz

| Etapa | O que acontece |
|---|---|
| 1 | Cria um frame bruto simulando câmera do drone |
| 2 | Recorta uma ROI (região de interesse) |
| 3 | Normaliza o frame para entrada no modelo |
| 4 | Cria anotações no formato YOLO |
| 5 | Extrai classes e bboxes das anotações |

```python
import numpy as np
```

## Etapa 1 — Frame bruto do drone

```python
# Frame bruto simulando imagem capturada pelo drone — 550x750 colorido
# dtype=uint8 → valores de 0 a 255 (formato padrão de imagem)
imagem_drone = np.random.randint(0, 256, size=(550, 750, 3), dtype=np.uint8)

print('Shape:', imagem_drone.shape)       # (550, 750, 3)
print('Dtype:', imagem_drone.dtype)       # uint8
print('Valor máximo:', imagem_drone.max()) # próximo de 255
print('Valor mínimo:', imagem_drone.min()) # próximo de 0
```

## Etapa 2 — Recorte da ROI (Region of Interest)

```python
# Recorta a região onde está o motociclista
# frame[y1:y2, x1:x2] — linhas 150 a 200, colunas 300 a 450
recorte = imagem_drone[150:200, 300:450]

print('Shape do recorte:', recorte.shape)  # (50, 150, 3)
```

## Etapa 3 — Normalização

```python
# Passo 1: converte de uint8 para float32
# Necessário antes de dividir — uint8 não suporta decimais
imagem_float = imagem_drone.astype(np.float32)

# Passo 2: normaliza — todos os pixels ficam entre 0.0 e 1.0
imagem_normalizada = imagem_float / 255.0

print('Dtype float:', imagem_float.dtype)                          # float32
print(f'Valor máximo normalizado: {imagem_normalizada.max():.4f}') # ~1.0
print(f'Valor mínimo normalizado: {imagem_normalizada.min():.4f}') # ~0.0
```

## Etapa 4 — Anotações no formato YOLO

```python
# Anotações dos motociclistas detectados no frame
# Formato: [classe, x_centro, y_centro, largura_box, altura_box]
# classe 0 = com capacete | classe 1 = sem capacete
anotacoes = np.array([
    [1, 0.20, 0.75, 0.36, 0.70],   # sem capacete
    [0, 0.50, 0.75, 0.36, 0.80],   # com capacete
    [1, 0.26, 0.95, 0.36, 0.20],   # sem capacete
    [0, 0.29, 0.35, 0.36, 0.60],   # com capacete
    [0, 0.35, 0.25, 0.36, 0.40],   # com capacete
    [0, 0.45, 0.50, 0.36, 0.30],   # com capacete
    [1, 0.85, 0.50, 0.36, 0.10],   # sem capacete
], dtype=np.float32)

print('Shape anotações:', anotacoes.shape)  # (7, 5)
```

## Etapa 5 — Extração de classes e bboxes

```python
# Coluna 0 → classes de todos os objetos
print('Classes:', anotacoes[:, 0])          # [1. 0. 1. 0. 0. 0. 1.]

# Colunas 1 e 2 → centros de cada bbox
print('Centros x:', anotacoes[:, 1])
print('Centros y:', anotacoes[:, 2])

# Colunas 1 a 4 → bboxes completas (x_centro, y_centro, largura_box, altura_box)
print('\nBboxes completas:\n', anotacoes[:, 1:5])
```

---
## Resumo do Bloco 1

| Conceito | Código | Para que serve no projeto |
|---|---|---|
| Criar frame | `np.random.randint(..., dtype=np.uint8)` | Simular imagem do drone |
| Recortar ROI | `frame[y1:y2, x1:x2]` | Isolar região do motociclista |
| Normalizar | `frame.astype(np.float32) / 255.0` | Preparar entrada para o modelo |
| Anotações YOLO | `np.array([[classe, x, y, w, h]])` | Representar detecções |
| Extrair classes | `anotacoes[:, 0]` | Saber quem tem/não tem capacete |
| Extrair bboxes | `anotacoes[:, 1:5]` | Saber onde está cada objeto |
