## Enunciado

Dado um frame de drone 720×1280 colorido:
1. Crie o frame com pixels aleatórios
2. Extraia os 3 canais RGB e imprima o shape de cada um
3. Recorte a região `[200:400, 300:700]` e imprima o shape
4. Converta a bbox abaixo para pixels e recorte a região:
   - `x_centro=0.52, y_centro=0.78, largura=0.25, altura=0.35`
5. Crie uma função reutilizável para converter bbox YOLO para pixels

```python
import numpy as np
```

## Resolução — Rubens Dutra

```python
# Questão 1 — Cria o frame 720x1280
# shape (720, 1280, 3) → altura=720, largura=1280, 3 canais RGB
imagem = np.random.randint(0, 256, size=(720, 1280, 3), dtype=np.uint8)
print("Shape do frame:", imagem.shape)
```

```python
# Questão 2 — Extrai os canais RGB
# [:, :, 0] → todas as linhas, todas as colunas, canal 0 (vermelho)
canal_red   = imagem[:, :, 0]   # canal vermelho
canal_green = imagem[:, :, 1]   # canal verde
canal_blue  = imagem[:, :, 2]   # canal azul

print("Canal R:", canal_red.shape)    # (720, 1280)
print("Canal G:", canal_green.shape)  # (720, 1280)
print("Canal B:", canal_blue.shape)   # (720, 1280)
```

```python
# Questão 3 — Recorte direto da região
# [200:400, 300:700] → linhas 200 a 400, colunas 300 a 700
roi = imagem[200:400, 300:700]
print("Shape do recorte:", roi.shape)  # (200, 400, 3)
```

```python
# Questão 4 — Converte bbox YOLO para pixels
# Importante: usar a resolução correta do frame criado (720x1280)
altura_frame  = 720    # mesma altura do frame criado
largura_frame = 1280   # mesma largura do frame criado

x_centro   = 0.52
y_centro   = 0.78
largura_box = 0.25
altura_box  = 0.35

# Fórmula padrão de conversão YOLO → pixels
x1 = int((x_centro - largura_box/2) * largura_frame)  # borda esquerda
y1 = int((y_centro - altura_box /2) * altura_frame)   # borda superior
x2 = int((x_centro + largura_box/2) * largura_frame)  # borda direita
y2 = int((y_centro + altura_box /2) * altura_frame)   # borda inferior

print(f"Coordenadas em pixels: x1={x1}, y1={y1}, x2={x2}, y2={y2}")

capacete = imagem[y1:y2, x1:x2]
print("Shape do capacete:", capacete.shape)
```

```python
# Questão 5 — Função reutilizável (criada pelo aluno)
# Escreve uma vez, usa para qualquer detecção do projeto
def bbox_para_pixels(x_centro, y_centro, largura_box, altura_box, largura_frame, altura_frame):
    x1 = int((x_centro - largura_box/2) * largura_frame)
    y1 = int((y_centro - altura_box /2) * altura_frame)
    x2 = int((x_centro + largura_box/2) * largura_frame)
    y2 = int((y_centro + altura_box /2) * altura_frame)
    return x1, y1, x2, y2  # retorna as 4 coordenadas em pixels

# Testando a função
x1, y1, x2, y2 = bbox_para_pixels(0.52, 0.78, 0.25, 0.35, 1280, 720)
recorte = imagem[y1:y2, x1:x2]
print(f"x1={x1}, y1={y1}, x2={x2}, y2={y2}")
print("Shape via função:", recorte.shape)
```
