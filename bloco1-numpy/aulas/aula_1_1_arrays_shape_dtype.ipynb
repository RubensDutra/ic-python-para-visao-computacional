## Tradução dos termos

| Termo em inglês | Tradução / Significado |
|---|---|
| **Array** | Arranjo — estrutura que guarda números organizados em dimensões |
| **Shape** | Forma — descreve quantas linhas, colunas e canais um array tem |
| **dtype** | *Data type* — tipo do dado — que tipo de número está guardado |
| **uint8** | Inteiro sem sinal de 8 bits — guarda valores de 0 a 255 (pixels) |
| **float32** | Número decimal de 32 bits — usado por redes neurais |
| **ndarray** | *N-dimensional array* — o objeto central do NumPy |
| **axis** | Eixo — cada direção de um array (altura, largura, canal) |

## Conceito

Toda imagem no computador é um array de números.  
Uma imagem colorida tem 3 dimensões: **(altura, largura, canais RGB)**.

Cada pixel carrega 3 valores ao mesmo tempo — R, G e B — que combinados formam uma cor.  
Exemplo: `[173, 42, 201]` = tom de roxo.

- `uint8` → formato da imagem bruta (0 a 255)
- `float32` → formato que o modelo espera receber (0.0 a 1.0)

```python
# Importa o NumPy — biblioteca de computação numérica
# Usamos o apelido 'np' por convenção universal
import numpy as np
```

## Parte A — Entendendo arrays de imagem

```python
# Cria uma imagem em escala de cinza simulada
# 4 linhas x 6 colunas — cada valor representa a intensidade de um pixel (0=preto, 255=branco)
# dtype=uint8 → tipo correto para pixels de imagem (valores de 0 a 255)
imagem_cinza = np.array([
    [  0,  50, 100, 150, 200, 255],
    [ 10,  60, 110, 160, 210, 245],
    [ 20,  70, 120, 170, 220, 235],
    [ 30,  80, 130, 180, 230, 225]
], dtype=np.uint8)

# shape → (linhas, colunas) → estrutura da imagem
print("Shape:", imagem_cinza.shape)   # (4, 6)

# dtype → tipo do dado armazenado
print("dtype:", imagem_cinza.dtype)   # uint8

# size → total de valores numéricos (linhas x colunas)
print("Total de valores (size):", imagem_cinza.size)  # 24
```

## Parte B — Imagem colorida (RGB)

```python
# Cria uma imagem colorida com pixels zerados (tela preta)
# np.zeros → preenche o array com zeros
# shape (4, 6, 3) → 4 linhas, 6 colunas, 3 canais de cor (R, G, B)
imagem_rgb = np.zeros((4, 6, 3), dtype=np.uint8)

print("Shape:", imagem_rgb.shape)   # (4, 6, 3)
print("dtype:", imagem_rgb.dtype)   # uint8

# Acessa o pixel na linha 0, coluna 0
# Retorna 3 valores: [R, G, B] — os três canais juntos formam a cor do pixel
print("Pixel [0, 0]:", imagem_rgb[0, 0])  # [0 0 0] = preto
```

## Parte C — Simulando um frame do drone

```python
# Simula um frame real capturado pelo drone
# np.random.randint → gera valores aleatórios de pixel entre 0 e 255
# size=(480, 640, 3) → resolução 480x640 com 3 canais RGB
# dtype=uint8 → tipo correto para imagem bruta
frame_drone = np.random.randint(0, 256, size=(480, 640, 3), dtype=np.uint8)

# Inspeciona o frame — isso é o que você vai fazer com cada imagem do projeto
print("Shape do frame:", frame_drone.shape)    # (480, 640, 3)
print("dtype:", frame_drone.dtype)             # uint8
print("Total de pixels:", frame_drone.shape[0] * frame_drone.shape[1])  # 307.200
print("Total de valores (size):", frame_drone.size)  # 921.600 (307.200 x 3)
print("Valor mínimo:", frame_drone.min())      # próximo de 0
print("Valor máximo:", frame_drone.max())      # próximo de 255
```
