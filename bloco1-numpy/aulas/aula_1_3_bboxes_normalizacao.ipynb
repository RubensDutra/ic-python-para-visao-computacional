# Aula 1.3 — Bounding Boxes e Normalização

**Bloco 1 — NumPy**  
**Projeto:** IC-FAPEMA — Detecção de capacetes em motociclistas via drone  
**Objetivo:** Normalizar imagens para entrada no modelo e trabalhar com anotações no formato YOLO

---

## Termos

| Termo | Conceito | Função no código |
|---|---|---|
| **Normalization** | Converter valores de 0-255 para 0.0-1.0 | `imagem / 255.0` |
| **dtype cast** | Converter o tipo do dado | `imagem.astype(np.float32)` |
| **Bounding Box** | Caixa que delimita o objeto detectado | `[x_centro, y_centro, largura, altura]` |
| **Annotation** | Arquivo que descreve onde está o objeto | arquivo `.txt` no formato YOLO |
| **Label** | Rótulo — classe do objeto detectado | `0` = com capacete, `1` = sem capacete |

---

## Conceito

**Por que normalizar?**  
Redes neurais trabalham melhor com números pequenos. Pixel `200` é grande demais — o modelo aprende mais devagar. Dividindo por 255, vira `0.78` — dentro de 0 e 1.

```
uint8   →  0 a 255    (imagem bruta do drone)
float32 →  0.0 a 1.0  (entrada no modelo)
```

**Formato YOLO de anotação:**
```
classe  x_centro  y_centro  largura  altura
  0      0.52      0.38      0.20     0.15
```
Todos os valores entre 0 e 1 — proporcionais ao tamanho da imagem.

````python
import numpy as np
````

## Parte A — Criação e inspeção da imagem bruta

````python
# Frame bruto do drone — 480x640 colorido
# dtype=uint8 → formato padrão de imagem, valores de 0 a 255
frame = np.random.randint(0, 256, size=(480, 640, 3), dtype=np.uint8)

print("dtype original:", frame.dtype)   # uint8
print("Valor máximo:", frame.max())     # próximo de 255
print("Valor mínimo:", frame.min())     # próximo de 0
````

## Parte B — Conversão de tipo e normalização

````python
# Passo 1: converte de uint8 para float32
# Necessário antes de dividir — uint8 não suporta decimais
frame_float = frame.astype(np.float32)

print("dtype convertido:", frame_float.dtype)  # float32
print("Valor máximo:", frame_float.max())       # 255.0 — ainda não normalizou
````

````python
# Passo 2: normaliza dividindo por 255
# Agora todos os pixels ficam entre 0.0 e 1.0
frame_normalizado = frame_float / 255.0

print("dtype normalizado:", frame_normalizado.dtype)  # float32
print("Valor máximo:", frame_normalizado.max())        # 1.0
print("Valor mínimo:", frame_normalizado.min())        # 0.0
````

## Parte C — Anotações no formato YOLO

````python
# Array de anotações — cada linha é um objeto detectado
# Formato: [classe, x_centro, y_centro, largura, altura]
# classe 0 = com capacete | classe 1 = sem capacete
anotacoes = np.array([
    [0, 0.52, 0.38, 0.20, 0.15],   # com capacete
    [0, 0.71, 0.45, 0.18, 0.12],   # com capacete
    [1, 0.30, 0.60, 0.16, 0.14],   # sem capacete
], dtype=np.float32)

print("Shape das anotações:", anotacoes.shape)  # (3, 5)

# Pegando só as classes — coluna 0
print("Classes:", anotacoes[:, 0])   # [0. 0. 1.]

# Pegando só os centros x e y — colunas 1 e 2
print("Centros (x, y):\n", anotacoes[:, 1:3])

# Pegando todas as bboxes — colunas 1 a 4
print("Bboxes completas:\n", anotacoes[:, 1:5])
````
