## 🧠 Conceito

Na Aula 4.1 você aprendeu a **carregar** uma imagem. Agora você vai aprender a **manipulá-la**.

Lembra do NumPy do Bloco 1? A imagem é um array NumPy — então tudo que você aprendeu de slicing se aplica aqui:

```
imagem.shape → (altura, largura, canais)
               linhas   colunas

Recortar:  imagem[y1:y2, x1:x2]
                  ↑ linhas  ↑ colunas
```

Isso é o **crop** — recortar exatamente onde está o motociclista, por exemplo.

A outra operação importante é converter para **escala de cinza**:  
Muitos algoritmos de visão computacional trabalham melhor sem cor — a imagem fica mais leve e o processamento é mais rápido.

```
Colorida:  shape (480, 640, 3)  → 3 canais (B, G, R)
Cinza:     shape (480, 640)     → 1 canal só
```

> No projeto, o crop será usado para **isolar o capacete** do restante da imagem do drone.

````python
# Importar bibliotecas
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

print("Bibliotecas carregadas!")
````

## 💻 Código — Parte 1: Carregar imagem

````python
# ── 1. Fazer upload e carregar a imagem ───────────────────────────────────────
print("Escolha uma foto de motociclista...")
uploaded = files.upload()
nome = list(uploaded.keys())[0]

imagem_bgr = cv2.imread(nome)                              # lê em BGR
imagem_rgb = cv2.cvtColor(imagem_bgr, cv2.COLOR_BGR2RGB)   # converte para exibir

altura, largura, canais = imagem_rgb.shape
print(f"Imagem carregada: {nome}")
print(f"Shape: {imagem_rgb.shape} → altura={altura}px, largura={largura}px, canais={canais}")
````

## 💻 Código — Parte 2: Recortar uma região (crop / ROI)

````python
# ── 2. Recortar a metade central da imagem ────────────────────────────────────
# Lembre: imagem[linhas, colunas] = imagem[y1:y2, x1:x2]

y1 = altura  // 4       # começa em 25% da altura
y2 = altura  * 3 // 4   # termina em 75% da altura
x1 = largura // 4       # começa em 25% da largura
x2 = largura * 3 // 4   # termina em 75% da largura

roi = imagem_rgb[y1:y2, x1:x2]   # recorte com slicing NumPy — igual ao Bloco 1!

print(f"Coordenadas do recorte: y={y1}:{y2}, x={x1}:{x2}")
print(f"Shape do recorte (ROI): {roi.shape}")

# Exibir original e recorte lado a lado
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

ax1.imshow(imagem_rgb)
ax1.set_title(f'Original\nShape: {imagem_rgb.shape}', fontsize=11)
ax1.axis('off')

ax2.imshow(roi)
ax2.set_title(f'Recorte (ROI)\nShape: {roi.shape}', fontsize=11)
ax2.axis('off')

fig.suptitle('crop — recortar região de interesse com slicing', fontsize=13, weight='bold')
plt.tight_layout()
plt.show()
````

## 💻 Código — Parte 3: Converter para escala de cinza

````python
# ── 3. Converter para escala de cinza ─────────────────────────────────────────
# Usamos imagem_bgr aqui pois cvtColor espera BGR → GRAY
imagem_cinza = cv2.cvtColor(imagem_bgr, cv2.COLOR_BGR2GRAY)

print(f"Shape colorida: {imagem_rgb.shape}   → 3 canais (B, G, R)")
print(f"Shape cinza:    {imagem_cinza.shape}  → 1 canal")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

ax1.imshow(imagem_rgb)
ax1.set_title(f'Colorida\nShape: {imagem_rgb.shape}', fontsize=11)
ax1.axis('off')

ax2.imshow(imagem_cinza, cmap='gray')   # cmap='gray' exibe em escala de cinza
ax2.set_title(f'Escala de cinza\nShape: {imagem_cinza.shape}', fontsize=11)
ax2.axis('off')

fig.suptitle('cvtColor — converter para grayscale', fontsize=13, weight='bold')
plt.tight_layout()
plt.show()
````

## 💻 Código — Parte 4: Separar os canais de cor

````python
# ── 4. Separar os 3 canais de cor ─────────────────────────────────────────────
# split() divide a imagem BGR em 3 arrays individuais
b, g, r = cv2.split(imagem_bgr)

print(f"Shape de cada canal: {b.shape}  → só 1 camada de cor")

fig, axs = plt.subplots(1, 4, figsize=(16, 4))

axs[0].imshow(imagem_rgb)
axs[0].set_title('Original')
axs[0].axis('off')

axs[1].imshow(r, cmap='Reds')     # canal vermelho
axs[1].set_title('Canal Vermelho')
axs[1].axis('off')

axs[2].imshow(g, cmap='Greens')   # canal verde
axs[2].set_title('Canal Verde')
axs[2].axis('off')

axs[3].imshow(b, cmap='Blues')    # canal azul
axs[3].set_title('Canal Azul')
axs[3].axis('off')

fig.suptitle('split() — canais de cor separados', fontsize=13, weight='bold')
plt.tight_layout()
plt.show()
````

````python
# ── 5. Juntar canais novamente com merge() ────────────────────────────────────
# merge() recebe uma lista com os 3 canais e reconstrói a imagem colorida
imagem_reconstruida_bgr = cv2.merge([b, g, r])                             # une em BGR
imagem_reconstruida_rgb = cv2.cvtColor(imagem_reconstruida_bgr, cv2.COLOR_BGR2RGB)  # converte

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
ax1.imshow(imagem_rgb)
ax1.set_title('Original')
ax1.axis('off')

ax2.imshow(imagem_reconstruida_rgb)
ax2.set_title('Reconstruída com merge()')
ax2.axis('off')

fig.suptitle('merge() — reunir canais separados', fontsize=13, weight='bold')
plt.tight_layout()
plt.show()

print("As duas imagens são idênticas — split + merge não altera nada!")
````
